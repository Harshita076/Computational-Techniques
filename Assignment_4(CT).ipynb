```python
# A village has several multiple mango gardens which has on an average 85% good quality mangoes. 
#On gaussian distribution basis. One of the person says his garden has 88% good quality mango 
#production. Assume σ=1, tell if his hypothesis should be accepted or rejected using the Z-score test. 
```

```python
# Given 
mean = 85          # population mean (%)
observed = 88      # sample value (%)
std_dev = 1        # standard deviation
significance = 0.05

# Calculate Z-score
z_score = (observed - mean) / std_dev

# Z critical value for 5% level (two-tailed test)
critical_value = 1.96

# Display result
print("Calculated Z-score:", z_score)

if abs(z_score) > critical_value:
    print("Null hypothesis is rejected")
    print("The person's claim is accepted")
else:
    print("Null hypothesis is accepted")
    print("The person's claim is rejected")
```
