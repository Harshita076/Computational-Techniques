```python
#2 fair and unbiased dice are rolled simultaneously to get some outcomes during the 
#experiment. Write a python program to compute number of outcomes and probability of getting 
#both faces same. 
```

```python
# Total outcomes when two dice are rolled
total_outcomes = 6 * 6

# Outcomes where both faces are the same
same_face_outcomes = 6

# Probability of getting both faces same
probability = same_face_outcomes / total_outcomes

print("Total number of outcomes:", total_outcomes)
print("Number of outcomes with both faces same:", same_face_outcomes)
print("Probability of getting both faces same:", probability)
```
